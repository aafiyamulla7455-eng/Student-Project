```python
import pandas as pd
import matplotlib.pyplot as plt

# Read dataset
student2 = pd.read_csv("student2.csv")

# Display dataset
print(student2)

# Find rows and columns
print(student2.shape)

# Dataset information
student2.info()

# Check missing values
print(student2.isnull().sum())

# Statistical values
print(student2.describe())

# Check duplicate values
print(student2.duplicated().sum())

# Remove duplicates
student2 = student2.drop_duplicates()

# Check data types
print(student2.dtypes)


# Gender-wise average marks
student2.groupby("Gender")["Marks"].mean().plot(
    kind="bar",
    ylabel="Average Marks",
    title="Gender Wise Average Marks"
)

plt.show()


# Course-wise average marks
student2.groupby("Course")["Marks"].mean().plot(
    kind="bar",
    ylabel="Average Marks",
    title="Course Wise Average Marks"
)

plt.show()


# City-wise average marks
student2.groupby("City")["Marks"].mean().plot(
    kind="bar",
    ylabel="Average Marks",
    title="City Wise Average Marks"
)

plt.show()


# Create Grade column
student2["Grade"] = pd.cut(
    student2["Marks"],
    bins=[0, 40, 60, 75, 90, 100],
    labels=["F", "C", "B", "A", "A+"]
)

# Display updated dataset
print(student2)


# Grade-wise number of students
student2["Grade"].value_counts().plot(
    kind="bar",
    ylabel="Number of Students",
    title="Grade Wise Students"
)

plt.show()
```
